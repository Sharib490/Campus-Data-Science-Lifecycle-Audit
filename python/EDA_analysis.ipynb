```python
gender_counts = df["Gender"].value_counts()

gender_counts
```

```python
plt.figure(figsize=(7, 5))

gender_counts.plot(
    kind="bar"
)

plt.title("Applicants by Gender")
plt.xlabel("Gender")
plt.ylabel("Number of Applicants")

plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python
category_counts = df["Category"].value_counts()

category_counts
```

```python
plt.figure(figsize=(8, 5))

category_counts.plot(
    kind="bar"
)

plt.title("Applicants by Category")
plt.xlabel("Category")
plt.ylabel("Number of Applicants")

plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python
city_counts = (
    df["City"]
    .value_counts()
    .sort_values(ascending=False)
)

city_counts
```

```python
plt.figure(figsize=(10, 6))

city_counts.plot(
    kind="bar"
)

plt.title("Applicants by City")
plt.xlabel("City")
plt.ylabel("Number of Applicants")

plt.xticks(rotation=45, ha="right")

plt.tight_layout()
plt.show()
```

```python
course_counts = (
    df["Course"]
    .value_counts()
    .sort_values(ascending=False)
)

course_counts
```

```python
plt.figure(figsize=(10, 6))

course_counts.plot(
    kind="bar"
)

plt.title("Course-wise Applicant Distribution")
plt.xlabel("Course")
plt.ylabel("Number of Applicants")

plt.xticks(rotation=45, ha="right")

plt.tight_layout()
plt.show()
```

```python
academic_columns = [
    "10th_Percentage",
    "12th_Percentage",
    "PCM_Percentage",
    "CET_Score",
    "CET_Percentile"
]

df[academic_columns].describe().T
```

```python
academic_average = (
    df[academic_columns]
    .mean()
    .round(2)
)

academic_average
```

```python
course_academic = (
    df.groupby("Course")[
        [
            "12th_Percentage",
            "PCM_Percentage",
            "CET_Percentile"
        ]
    ]
    .mean()
    .round(2)
)

course_academic
```

```python
eligibility_counts = (
    df["Eligibility_Status"]
    .value_counts()
)

eligibility_counts
```

```python
plt.figure(figsize=(7, 5))

eligibility_counts.plot(
    kind="bar"
)

plt.title("Student Eligibility Status")
plt.xlabel("Eligibility Status")
plt.ylabel("Number of Students")

plt.xticks(rotation=0)

plt.tight_layout()
plt.show()
```

```python
eligibility_course = pd.crosstab(
    df["Course"],
    df["Eligibility_Status"]
)

eligibility_course
```

```python
eligibility_course.plot(
    kind="bar",
    figsize=(10, 6)
)

plt.title("Eligibility Status by Course")
plt.xlabel("Course")
plt.ylabel("Number of Students")

plt.xticks(rotation=45, ha="right")

plt.tight_layout()
plt.show()
```

```python
course_admission_rate = (
    df.groupby("Course")
    .agg(
        Total_Applicants=("Student_ID", "count"),
        Confirmed_Admissions=(
            "Admission_Status",
            lambda x: (x == "Confirmed").sum()
        )
    )
)

course_admission_rate["Admission_Rate_%"] = (
    course_admission_rate[
        "Confirmed_Admissions"
    ]
    /
    course_admission_rate[
        "Total_Applicants"
    ]
    * 100
).round(2)

course_admission_rate
```
